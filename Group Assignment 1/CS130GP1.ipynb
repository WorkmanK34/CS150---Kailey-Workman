```python
from random import randint
class GuessTooLongError(Exception):
    pass
class InvalidGuessError(Exception):
    pass
flag = True
win_counter = 0 
loss_counter = 0 
```

```python
# Target Sequence Generation 
class SequenceGenerator():
    def __init__(self):
        pass
    def generate(self):
        generated_num = []
        generated_num.append(randint(1,9))
        for i in range(3):
            generated_num.append(randint(0,9))
        return generated_num
        
```

```python
# User input number storage 
class UserGuess(SequenceGenerator):
    def guess(self):
        while True:
            try:
                guess = input('Enter Your Four Digit Number!')
                if len(guess) != 4:
                    raise GuessTooLongError
                if not guess.isdigit():
                    raise InvalidGuessError
            except InvalidGuessError:
                print('Make Sure You Are Entering An Integer Value')
            except GuessTooLongError:
                print('Please Enter A FOUR Digit Guess')
            else:
                guessed_num = []
                for n in guess:
                    guessed_num.append(int(n))
                return guessed_num
            
```

```python
while flag == True:
    print(f'Welcome to Mastermind. Can you guess the four digit number in 10 guesses? Your current record is {win_counter} wins and {loss_counter} losses.')
    target_sequence = SequenceGenerator()
    target = target_sequence.generate()
    guess_counter = 0 
    won = False
    while guess_counter < 10:
        users_guess = UserGuess()
        guess = users_guess.guess()
        num_present = 0
        num_correct = 0
        readable_target = target.copy()
        readable_guess = guess.copy()
        for i in range(4):
            if readable_guess[i] == readable_target[i]:
                num_correct += 1
                readable_guess[i] = None
                readable_target[i] = None
        for i in range(4):
            if readable_guess[i] is not None and readable_guess[i] in readable_target:
                num_present += 1
                readable_target.remove(readable_guess[i])
        print(f'{num_correct} digit(s) is/are correct and in the correct position\n{num_present} digit(s) is/are correct but in the wrong position.')
        if num_correct == 4:
            print('You correctly guessed the number!')
            win_counter += 1
            won = True
            break
        guess_counter += 1
    if not won:
        print(f'You Lost! The Number Was {target}')
        loss_counter += 1
    print(f'Your current record is {win_counter} wins and {loss_counter} losses.')
    play_again = input("Do You Want To Play Again? Type y For Yes And n For No.").lower()
    if play_again == 'n':
        flag = False
    

        
```
