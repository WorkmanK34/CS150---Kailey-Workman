```python
# Task 2 
dic = {'red':'rouge','orange':'orange','yellow':'jaune','green':'vert','blue':'bleu','purple':'violet','black':'noir','white':'blanc','brown':'marron','gray':'gris'}
flag = True
while flag == True:
    inp = input('Enter an English colour for translation to French. Press Enter to stop').strip().lower()
    if inp == '':
        flag = False   
    elif inp not in dic:
        print(f'The French word for {inp} is not in this dictionary.')
    else:
        print(f'The French word for {inp} is {dic[inp]}')
```

```python
# Task 3 
avtemps = {'January':[43,26],'February':[48,30],'March':[58,39],'April':[71,49],'May':[79,61],'June':[87,68],'July':[90,71],'August':[88,70],'September':[83,63],'October':[71,51],'November':[57,38],'December':[49,33]}
while True:
    mon = input('Enter a month and I will tell you the high and low temperature. Enter Quit to end.').strip().title()
    if mon == 'Quit':
        break 
    elif mon not in avtemps:
        print('That is not a month')
    else:
        print(f'The high for {mon} was {avtemps[mon][0]} and the low was {avtemps[mon][1]}')
```

```python
# Task 4 
inventory = {"hammer":[15, 12.49], "wrench":[20, 7.99], "screwdriver": [40, 3.99], "nailgun":[10, 36.59], "padlock":[25, 5.99], "roll_of_duct_tape": [30, 4.95]}
while True:    
    choice = int(input("Welcome to the store.\n" 
                       "To view the stock of an item enter 1. \n"
                       "To view items in your price range enter 2. \n"
                       "To view the total quantity of inventory enter 3. \n"
                       "To view the total price of the inventory enter 4. \n"
                      "To exit enter 0"))
    if choice == 0:
        break
    elif choice == 1:
        item = input("Enter the name of the item you wish to know the stock of.").strip().lower()
        if item in inventory:
            print(f"There are {inventory[item][0]} {item}s")
        else:
            print("That item is not in stock here.")
    elif choice == 2:
        budget = float(input("Enter your budget."))
        can_buy = []
        for i in inventory:
            if inventory[i][1] <= budget:
                can_buy.append(i)
        if not can_buy:
            print('Nothing here is in your budget')
        else:
            print(f'You can buy these items: {can_buy}')
    elif choice == 3:
        total_quantity = 0 
        for item in inventory:
            total_quantity += inventory[item][0]
        print(f'The total quantity of all inventory is {total_quantity} items')
    elif choice == 4:
        total_value = 0 
        for item in inventory:
            total_value += inventory[item][0]*inventory[item][1]
        print(f'The total value of all the inventory is ${total_value}')
    else:
        print('That was not a valid response.')
        
```
