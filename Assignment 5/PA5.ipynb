```python
#Task 1 
sat_in = True
sun_in = True
actSat = set()
actSun = set()
while sat_in:
    activity = input('Enter the activities you do on a Saturday. Type done when you are finished.').lower()
    if activity == 'done':
        sat_in = False
    else:
        actSat.add(activity) 
while sun_in:
    activity = input('Enter the activities you do on a Sunday. Type done when you are finished.').lower()
    if activity == 'done':
        sun_in = False
    else:
        actSun.add(activity)
print(actSat)
print(actSun)
print(actSun.intersection(actSat))
print(actSun.union(actSat))
```

```python
#Task 2
def triangle_solve(leg1, leg2):
    parameters = ()
    hypo = (leg1**2 + leg2**2 )**0.5
    parameters = parameters+(hypo,) 
    perimeter = leg1 + leg2 + hypo
    s = perimeter / 2
    area = (s*(s-leg1)*(s-leg2)*(s-hypo))**0.5
    parameters = parameters+(perimeter,)
    parameters = parameters+(area,)
    print(parameters)
```

```python
triangle_solve(3,4)
```

```python
#Task 3
class Calculator():
    def __init__(self,a,b):
        self.a = a 
        self.b = b 
        self.prev_result = 0
        self.total_calculations = 0 
    def add(self):
        result = self.a + self.b
        self.total_calculations += 1
        self.prev_result = result
        return result
    def subtract(self):
        result = self.a - self.b
        self.total_calculations += 1
        self.prev_result = result
        return result
    def multiply(self):
        result =  self.a * self.b
        self.total_calculations += 1
        self.prev_result = result
        return result
    def divide(self):
        try:
            result = self.a / self.b
        except ZeroDivisionError:
            print('You cannot divide by zero')
            return None
        self.total_calculations += 1
        self.prev_result = result
        return result
    def power(self):
        result =  self.a ** self.b
        self.total_calculations += 1
        self.prev_result = result
        return result
    def info(self):
        print(f'The previous result was {self.prev_result} and the total number of calculations performed is {self.total_calculations}.')
```

```python
newCal = Calculator(6,3)
print(newCal.add())
print(newCal.subtract())
print(newCal.multiply())
print(newCal.divide())
print(newCal.power())
newCal.info()
```
