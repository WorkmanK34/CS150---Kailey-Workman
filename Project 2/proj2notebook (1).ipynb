```python
# Task 1 - Finding and Uploading a csv file 
import pandas as pd #importing pandas for use 
df = pd.read_csv('Dengue.csv') # df is a variable that now contains my dataset
#print(df.head())
```

```python
#White Blood Cell Count Statistics 
#This section calculates basic statistics for the wbc_count column 
print(df["wbc_count"].mean())
print(df["wbc_count"].max())
print(df["wbc_count"].min())
print(df["wbc_count"].std())
print(df["wbc_count"].median())
print((df["wbc_count"].max())-df["wbc_count"].min()) #there is no range fxn so we do max minus min to find the range 
```

The white blood cell count has a mean of 4338.03 cells per microliter
The maximum value is 10900 cells/microliter and the minimum is 2000 cells/microliter, giving a range of 8900
The standard deviation is 2344.53, which is substantial 
The median is 8900 cells/microliter 

```python
#Platelet Count Statistics
#This section calculates basic statistics for the platelet_count column 
print(df['platelet_count'].mean())
print(df['platelet_count'].max())
print(df['platelet_count'].min())
print(df['platelet_count'].std())
print(df['platelet_count'].median())
print((df['platelet_count'].max())-df['platelet_count'].min())
```

```python
# Task 3 - Scatter Plot
c1 = df.plot.scatter(x='wbc_count',y='platelet_count')
c1.set_title("Platelet to WBC Cell Ratio (PWR)")
```
