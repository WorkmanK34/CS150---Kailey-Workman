```python
# task 1
side1 = float(input('Enter the length of the first side.'))
side2 = float(input('Enter the length of the second side.'))
side3 = float(input('Enter the length of the third side.'))
if side1 == side2:
    if side1 == side3:
        triangletype = 'an equilateral'
    else:
        triangletype = 'an isosceles'
elif side1 == side3:
    triangletype = 'an isosceles'
else:
    triangletype = 'a scalene'
print('You provided the side lengths of ' + triangletype + ' triangle.')
```

```python
# task 2
student_grades = []
counter = 0 
atriskcounter = 0 
total = 0
while counter <= 4:
    grade = float(input('Enter the students grade'))
    student_grades.append(grade)
    total = total + grade
    if grade < 65:
        atriskcounter += 1 
    counter += 1
print(student_grades)   
print('The class average is a '+ str(total/5) + '%.')
print('The minimum grade is a ' + str(min(student_grades)) + '%.')
print('The maximum grade is a ' + str(max(student_grades)) + '%.')
print(str(atriskcounter) + ' students are at risk in the course')
```

```python
# task 3 
mult_of_2 = []
number = 1
while number <= 100: 
    if number%2 == 0:
        mult_of_2.append(number)
    number += 1
print(mult_of_2)

mult_of_3 = []
value = 1 
while value <= 100:
    if value%3 == 0:
        mult_of_3.append(value) 
    value += 1
print(mult_of_3)

mult_of_both = []
for num in mult_of_2:
    if num in mult_of_3:
        mult_of_both.append(num)
print(mult_of_both)
    
mult_of_either = mult_of_2.copy()
for enum in mult_of_3:
    if enum not in mult_of_either:
        mult_of_either.append(enum)
mult_of_either.sort()
print(mult_of_either)
```

```python
# task 4 
shopping_list = []

while len(shopping_list) < 7:
    item = input('Enter an item to add to your shopping list. Enter done when you are finished. ')
    if item.lower() == 'done':
        break
    elif item in shopping_list:
        print('You have already added this item to the list')
    else:
        shopping_list.append(item)
print(shopping_list) 
print('There are ' + str(len(shopping_list))+ ' items on your list.')

num = len(shopping_list) -1 
while num >= 0:
    print(shopping_list[num])
    num = num - 1
alph_s_list = sorted(shopping_list)
print(alph_s_list)
    
```
